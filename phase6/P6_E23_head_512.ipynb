checkpoint in Drive, then prompts you to upload the checkpoint from your
computer. It downloads the official dataset automatically. Set a T4 GPU and
run cells in order. Keep the scored baseline submission until this candidate
receives a better Codabench score. No score improvement is guaranteed.

original snapshot. This dense experiment does not fulfill the course's

```python
from capstone.phase1.src.model.config import LegalLMConfig
from capstone.phase1.src.model.legal_lm import LegalLM

tokenizer = Tokenizer.from_file(str(TOKENIZER_PATH))
checkpoint = torch.load(LOCAL_CHECKPOINT, map_location="cpu", mmap=True, weights_only=False)
assert "model" in checkpoint and "run_config" in checkpoint
config = LegalLMConfig(**checkpoint["run_config"]["model_config"])
assert config.vocab_size == tokenizer.get_vocab_size()
model = LegalLM(config)
model.load_state_dict(checkpoint["model"], strict=True, assign=True)
del checkpoint
if METHOD == "last_four_blocks":
    for parameter in model.parameters():
        parameter.requires_grad_(False)
    for block in model.blocks[-4:]:
        for parameter in block.parameters():
            parameter.requires_grad_(True)
    for parameter in model.final_norm.parameters():
        parameter.requires_grad_(True)
model.to("cuda")
model.train()
torch.cuda.empty_cache()
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print("Method:", METHOD, "trainable parameters:", trainable)
```

```python
BOS = config.bos_token_id
EOS = config.eos_token_id
PAD = config.pad_token_id
PROMPT_A = tokenizer.encode("Summarize the following legal judgment:\n", add_special_tokens=False).ids
PROMPT_B = tokenizer.encode("\nSummary:\n", add_special_tokens=False).ids
MAX_DOC_TOKENS = 1024
MAX_SUMMARY_TOKENS = 512
MAX_INPUT_TOKENS = 1600

def document_ids(document):
    tokens = tokenizer.encode(document, add_special_tokens=False).ids
    if len(tokens) > MAX_DOC_TOKENS:
        tokens = tokens[:768] + tokens[-256:]
    return tokens

def prefix_ids(document):
    result = [BOS] + PROMPT_A + document_ids(document) + PROMPT_B
    assert len(result) + MAX_SUMMARY_TOKENS <= MAX_INPUT_TOKENS
    return result

class SummaryDataset(Dataset):
    def __init__(self, rows):
        self.examples = []
        for row in rows:
            prefix = prefix_ids(row["document"])
            target = tokenizer.encode(row["summary"], add_special_tokens=False).ids[:MAX_SUMMARY_TOKENS-1] + [EOS]
            ids = prefix + target
            labels = [-100] * len(prefix) + target
            self.examples.append((ids, labels))
    def __len__(self): return len(self.examples)
    def __getitem__(self, index):
        ids, labels = self.examples[index]
        return (torch.tensor(ids, dtype=torch.long),
                torch.tensor(labels, dtype=torch.long))

# Keep the baseline's held-out split so runs can be compared fairly.
selection = np.random.default_rng(781).permutation(len(train_rows))
dev_indices = set(int(x) for x in selection[:350])
train_set = SummaryDataset([r for i, r in enumerate(train_rows) if i not in dev_indices])
dev_set = SummaryDataset([r for i, r in enumerate(train_rows) if i in dev_indices])
train_loader = DataLoader(train_set, batch_size=1, shuffle=True, num_workers=0, pin_memory=True)
dev_loader = DataLoader(dev_set, batch_size=1, shuffle=False, num_workers=0, pin_memory=True)
print("Tokenized:", len(train_set), "training and", len(dev_set), "held-out rows")
```

```python
optimizer = torch.optim.AdamW((p for p in model.parameters() if p.requires_grad),
                               lr=LEARNING_RATE, weight_decay=0.01)
scaler = torch.amp.GradScaler("cuda")
MAX_UPDATES = 450
GRAD_ACCUM = 4
FIRST_EXPORT_UPDATE = 80
TRAIN_BUDGET_MINUTES = 150
best_loss = float("inf")
best_update = 0
started = time.monotonic()
history = []

@torch.inference_mode()
def dev_loss(limit=40):
    model.eval()
    losses = []
    for i, (ids, labels) in enumerate(dev_loader):
        if i >= limit: break
        with torch.amp.autocast("cuda", dtype=torch.float16):
            loss = model(ids.to("cuda"), labels=labels.to("cuda")).loss
        losses.append(float(loss))
    return float(np.mean(losses))

@torch.inference_mode()
def generate_summary(document):
    model.eval()
    prefix = prefix_ids(document)
    ids = torch.tensor([prefix], device="cuda", dtype=torch.long)
    with torch.amp.autocast("cuda", dtype=torch.float16):
        generated = model.generate(ids, max_new_tokens=MAX_SUMMARY_TOKENS,
                                   eos_token_id=EOS, pad_token_id=PAD,
                                   do_sample=False, repetition_penalty=1.08)
    continuation = generated[0, len(prefix):].tolist()
    if EOS in continuation:
        continuation = continuation[:continuation.index(EOS)]
    summary = tokenizer.decode(continuation, skip_special_tokens=True).strip()
    return summary if len(summary) >= 20 else document[:900].strip()

def copy_atomic(source, target):
    temporary = target.with_name(target.name + ".uploading")
    shutil.copy2(source, temporary)
    os.replace(temporary, target)

def export_current(update):
    global best_loss, best_update
    loss = dev_loss()
    print("Update", update, "held-out loss", loss, flush=True)
    if loss >= best_loss and (DRIVE_OUTPUT / "submission.zip").exists():
        return
    predictions = []
    fallback_count = 0
    for i, row in enumerate(test_rows, 1):
        summary = generate_summary(row["document"])
        if summary == row["document"][:900].strip(): fallback_count += 1
        predictions.append({"id": str(row["id"]), "summary": summary})
        if i % 10 == 0: print("Generated", i, "of", len(test_rows), flush=True)
    assert len(predictions) == 100
    assert [p["id"] for p in predictions] == [str(r["id"]) for r in test_rows]
    assert all(isinstance(p["summary"], str) and p["summary"] for p in predictions)
    local_json = LOCAL / "prediction.json"
    local_zip = LOCAL / "submission.zip"
    local_json.write_text(json.dumps(predictions, ensure_ascii=False, separators=(",", ":")) + "\n", encoding="utf-8")
    with zipfile.ZipFile(local_zip, "w", zipfile.ZIP_DEFLATED, compresslevel=9) as archive:
        archive.write(local_json, "prediction.json")
    with zipfile.ZipFile(local_zip) as archive:
        assert archive.namelist() == ["prediction.json"] and archive.testzip() is None
    manifest = {"run_id": RUN_ID, "method": METHOD, "parent": CHECKPOINT_NAME,
                "parent_kind": "Phase 6 fine-tuned descendant of student's Phase 4 dense LegalLM",
                "original_phase1_snapshot": False, "dataset_revision": DATASET_REVISION,
                "train_rows": len(train_set), "dev_rows": len(dev_set), "test_rows": len(test_rows),
                "updates": update, "held_out_loss": loss, "fallback_summaries": fallback_count,
                "zip_sha256": hashlib.sha256(local_zip.read_bytes()).hexdigest()}
    local_manifest = LOCAL / "manifest.json"
    local_manifest.write_text(json.dumps(manifest, indent=2) + "\n", encoding="utf-8")
    for source in (local_json, local_zip, local_manifest):
        copy_atomic(source, DRIVE_OUTPUT / source.name)
    try:
        local_state = LOCAL / "fine_tuned_state.pt"
        torch.save({"model": {k:v.detach().cpu() for k,v in model.state_dict().items()},
                    "run_config": {"model_config": vars(config)},
                    "manifest": manifest}, local_state)
        copy_atomic(local_state, DRIVE_OUTPUT / local_state.name)
    except Exception as exc:
        print("ZIP saved; checkpoint copy failed:", repr(exc), flush=True)
    best_loss, best_update = loss, update
    print("READY:", DRIVE_OUTPUT / "submission.zip", flush=True)

model.train()
optimizer.zero_grad(set_to_none=True)
update = 0
for step, (ids, labels) in enumerate(train_loader, 1):
    with torch.amp.autocast("cuda", dtype=torch.float16):
        loss = model(ids.to("cuda", non_blocking=True),
                     labels=labels.to("cuda", non_blocking=True)).loss
    scaler.scale(loss / GRAD_ACCUM).backward()
    if step % GRAD_ACCUM:
        continue
    scaler.unscale_(optimizer)
    torch.nn.utils.clip_grad_norm_((p for p in model.parameters() if p.requires_grad), 1.0)
    scaler.step(optimizer); scaler.update()
    optimizer.zero_grad(set_to_none=True)
    update += 1
    if update <= 30:
        lr = LEARNING_RATE * update / 30
    else:
        lr = LEARNING_RATE * max(0.1, (MAX_UPDATES-update)/(MAX_UPDATES-30))
    for group in optimizer.param_groups: group["lr"] = lr
    if update % 25 == 0:
        elapsed = (time.monotonic()-started)/60
        print(f"update {update}/{MAX_UPDATES} loss {float(loss):.4f} elapsed {elapsed:.1f}m", flush=True)
    if update == FIRST_EXPORT_UPDATE:
        export_current(update)
        model.train()
    if update >= MAX_UPDATES or (time.monotonic()-started)/60 >= TRAIN_BUDGET_MINUTES:
        break
if update > best_update:
    export_current(update)
print("Best ZIP:", DRIVE_OUTPUT / "submission.zip", "held-out loss", best_loss)
```
